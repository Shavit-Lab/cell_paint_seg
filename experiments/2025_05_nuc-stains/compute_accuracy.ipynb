# Compare

```python
import numpy as np
from collections import defaultdict

def evaluate_label_overlap(pred, truth):
    assert pred.shape == truth.shape

    pred_labels = np.unique(pred)
    pred_labels = pred_labels[pred_labels != 0]  # Exclude background

    truth_labels = np.unique(truth)
    truth_labels = truth_labels[truth_labels != 0]  # Exclude background

    # Mapping from predicted label → overlapping truth labels
    pred_to_truth = {}
    truth_covered = set()

    for pl in pred_labels:
        pred_mask = pred == pl
        overlapping_truth_labels = np.unique(truth[pred_mask])
        overlapping_truth_labels = overlapping_truth_labels[overlapping_truth_labels != 0]

        # Assert each predicted label overlaps with at most one truth label (or 0)
        assert len(overlapping_truth_labels) <= 1, (
            f"Predicted label {pl} overlaps with multiple truth labels: {overlapping_truth_labels}"
        )

        if len(overlapping_truth_labels) == 1:
            tl = overlapping_truth_labels[0]
            pred_to_truth[pl] = tl
            truth_covered.add(tl)
        else:
            pred_to_truth[pl] = 0  # Overlaps only with background

    # Count TPs, FPs, FNs
    tp = sum(1 for val in pred_to_truth.values() if val != 0)
    fp = sum(1 for val in pred_to_truth.values() if val == 0)
    fn = len(set(truth_labels) - truth_covered)

    return tp, fp, fn
```

## Sanity check that all nucs have perfect overlap (and FN=0) from c9 segmentation

```python
# for every key that file_dict_manual and file_dict_pipeline_all have in common, plot the two images side by side
for key in file_dict_manual.keys() & file_dict_pipeline_all.keys():
    manual_image = io.imread(file_dict_manual[key])
    pipeline_image = io.imread(file_dict_pipeline_all[key])

    tp, fp, fn = evaluate_label_overlap(pipeline_image, manual_image)
    
    # binarize each image 
    manual_image = (manual_image > 0).astype(int)
    pipeline_image = (pipeline_image > 0).astype(int)

    # Make an image that is black where both images are 0, blue where manual is 1 and pipeline is 0, red where manual is 0 and pipeline is 1, and green where both are 1
    green_channel = (manual_image & pipeline_image).astype(int) * 255
    red_channel = ((~manual_image) & pipeline_image).astype(int) * 255
    blue_channel = (manual_image & ~pipeline_image).astype(int) * 255
    combined_image = np.stack((red_channel, green_channel, blue_channel), axis=-1)
    plt.figure(figsize=(10, 5))
    plt.imshow(combined_image)
    plt.title(f"Image {key}. Annotated Alive {tp} Annotated Dead {fp} Problematic {fn}")
    plt.axis('off')
    plt.show()
    
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np
from skimage import io

# Initialize totals
total_tp, total_fp, total_fn = 0, 0, 0

# Prepare a 2x6 grid
fig, axes = plt.subplots(2, 6, figsize=(18, 6), dpi=300)
axes = axes.flatten()

# Shared legend handles
red_patch = mpatches.Patch(color='red', label='False Positive')
green_patch = mpatches.Patch(color='green', label='True Positive')
blue_patch = mpatches.Patch(color='blue', label='False Negative')

# Loop through common keys and display up to 12
for idx, key in enumerate(file_dict_manual.keys() & file_dict_pipeline.keys()):
    if idx >= 12:
        break  # Only show 12 images

    manual_image = io.imread(file_dict_manual[key]).T
    pipeline_image = io.imread(file_dict_pipeline[key]).T
    pipeline_image_all = io.imread(file_dict_pipeline_all[key]).T
    pipeline_image_all[pipeline_image > 0] = 0  
    pipeline_image = pipeline_image_all
    # manual_image = np.flip(manual_image, axis=0)  # Flip the manual image vertically
    # manual_image = np.flip(manual_image, axis=1)
    # pipeline_image = np.flip(pipeline_image, axis=0)  # Flip the pipeline image vertically
    # pipeline_image = np.flip(pipeline_image, axis=1)
    

    tp, fp, fn = evaluate_label_overlap(pipeline_image, manual_image)

    total_tp += tp
    total_fp += fp
    total_fn += fn

    # Binarize
    manual_bin = (manual_image > 0).astype(int)
    pipeline_bin = (pipeline_image > 0).astype(int)

    # RGB channels
    green = (manual_bin & pipeline_bin).astype(np.uint8) * 255
    red = ((~manual_bin.astype(bool)) & pipeline_bin.astype(bool)).astype(np.uint8) * 255
    blue = (manual_bin.astype(bool) & ~pipeline_bin.astype(bool)).astype(np.uint8) * 255

    combined = np.stack((red, green, blue), axis=-1)

    ax = axes[idx]
    ax.imshow(combined)
    ax.set_title(f"{key}\nTP {tp} FP {fp} FN {fn}", fontsize=8)
    ax.axis('off')

# Hide any unused axes if < 12 images
for i in range(idx + 1, len(axes)):
    axes[i].axis('off')

# Add a single legend outside the grid
fig.legend(
    handles=[red_patch, green_patch, blue_patch],
    loc='center right',
    bbox_to_anchor=(1.02, 0.5),
    framealpha=1
)

total_recall = total_tp / (total_tp + total_fn)
total_prec = total_tp / (total_tp + total_fp)
plt.suptitle(f"Recall: {total_recall:.2f} Prec: {total_prec:.2f} (TP: {total_tp}, FP: {total_fp}, FN: {total_fn})")

plt.tight_layout()
plt.subplots_adjust(right=0.9)  # Make space for legend
#plt.savefig("pipeline_acc.png")
#plt.show()

```
